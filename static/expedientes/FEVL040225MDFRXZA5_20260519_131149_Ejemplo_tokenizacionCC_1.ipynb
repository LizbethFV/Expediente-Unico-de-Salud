Tokenización
La tokenización es un paso crucial en las tareas de PLN (Procesamiento del Lenguaje Natural), donde el texto se divide en unidades más pequeñas llamadas tokens. Estos tokens pueden variar desde caracteres individuales hasta oraciones completas, aunque suelen ser palabras o subpalabras.

Lemantización
En el campo del procesamiento del lenguaje natural (PLN), la lematización es el proceso de reducir las palabras a su forma base, conocida como lema. Esto ayuda a agrupar palabras similares considerando su significado y estructura gramatical. De esta manera, se garantiza que las palabras con significados similares reciban el mismo tratamiento.

**Lematización**

La lematización es un proceso lingüístico y computacional que reduce las palabras flexionadas (plurales, conjugaciones, etc.) a su forma base o "lema" (la palabra que aparece en el diccionario) utilizando análisis morfológico y contextos. Es esencial en el procesamiento del lenguaje natural (PLN) para normalizar texto.

```python
import spacy
```

```python
import spacy.cli
spacy.cli.download("en_core_web_sm")
```

```python
nlp = spacy.load("en_core_web_sm")
```

```python
def lemmantize_text (text):
  doc = nlp(text)
  return ' '.join([token.lemma_ for token in doc])
```

```python
df['output_lemmatized'] = df['output'].apply(lemmantize_text)
df['output_lemmatized']

```

```python
df['output_lemmatized'][3]
```

```python
X = df['output_lemmatized']
y = df['specialty']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
vectorizer = CountVectorizer()
X_train_transformed = vectorizer.fit_transform(X_train)
X_test_transformed = vectorizer.transform(X_test)
model = MultinomialNB()
model.fit(X_train_transformed, y_train)
y_pred = model.predict(X_test_transformed)
print("Accuracy:", metrics.accuracy_score(y_test, y_pred))
```
